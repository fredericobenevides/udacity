```python
# importing packages
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

%matplotlib inline
```

```python
# initializing values
x = [5]
y = [29]
```

```python
def create_line(x_array, w1, w2, p=0, lr = 1):
    y = []
    for x in x_array:
        #calc = w1*x + w2
        calc = (w1+(p*lr))*x + (w2*lr)
        y.append(calc)
    
    return y
```

```python
fig = plt.figure()

x_line = [0,2,4,6,8,10,12,14]
y_line = create_line(x_line, w1=1, w2=1)
print(y_line)

plt.scatter(x, y)
plt.plot(x_line, y_line)
plt.title('Using absolute trick')
```
