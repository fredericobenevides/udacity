# Cleaning the data
Cleaning data to remove duplicates, null values,

```python
import pandas as pd
df = pd.read_csv('cancer_data_means.csv')

# check which columns have missing values with info()
df.info()
```

```python
# Let's fill na with the mean of the columns
mean = df['smoothness_mean'].mean()
df['smoothness_mean'].fillna(mean, inplace=True)

mean = df['texture_mean'].mean()
df['texture_mean'].fillna(mean, inplace=True)

mean = df['symmetry_mean'].mean()
df['symmetry_mean'].fillna(mean, inplace=True)

# confirm if NA was filled
df.info()
```

```python
# Verify duplicated values
df.duplicated().sum()
```

```python
# Remove duplicated
df.drop_duplicates(inplace=True)

# Verify if is dropped
df.duplicated().sum()
```

# Visualizing Malignment x Benign

```python
import pandas as pd
% matplotlib inline

df = pd.read_csv('cancer_data_edited.csv')
df.head()
```

```python
# Create dataframe for Malignment
df_m = df[df['diagnosis'] == 'M']
df_m.head()
```

```python
# Create dataframe for Benign
df_b = df[df['diagnosis'] == 'B']
df_b.head()
```

```python
df_m['area'].describe()
```

```python
df_b['area'].describe()
```

```python
'The mean area of malignant tumors is {0:.4f} while that of benign tumors is {1:.4f}'.format(df_m['area'].mean(), df_b['area'].mean())
```

```python
import matplotlib.pyplot as plt
% matplotlib inline

fig, ax = plt.subplots(figsize=(8, 6))
ax.hist(df_b['area'], alpha=0.5, label='benign')
ax.hist(df_m['area'], alpha=0.5, label='malignant')
ax.set_title('Distributions of Benign and Malignant Tumor Areas')
ax.set_xlabel('Area')
ax.set_ylabel('Count')
ax.legend(loc='upper right')
plt.show()
```
